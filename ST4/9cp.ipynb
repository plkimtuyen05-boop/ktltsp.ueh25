```python
from agentic_vnstock import AgenticVNStock
import pandas as pd
import time

client = AgenticVNStock()

tickers = ["VCB", "TCB", "BID", "VNM", "MSN", "SAB", "VHM", "NVL", "KDH"]

all_data = []

for ticker in tickers:
    try:
        print(f"\n===== {ticker} =====")
        
        df = client.get_stock_historical(ticker, "2018-01-01", "2026-04-21")
        
        df["ticker"] = ticker
        
        # 👉 In ra dữ liệu
        print(df.to_string())   # in full
        
        # nếu dữ liệu quá dài, bạn nên dùng:
        # print(df.head())  # 5 dòng đầu
        # print(df.tail())  # 5 dòng cuối
        
        all_data.append(df)
        
        time.sleep(1)
        
    except Exception as e:
        print(f"Error with {ticker}: {e}")

# Gộp và lưu
final_df = pd.concat(all_data, ignore_index=True)
final_df.to_csv("portfolio_9stocks.csv", index=False)

print("\nDone! Data saved.")
```

```python
# kiểm tra trùng
duplicates = final_df.duplicated(subset=["Date", "ticker"])
print("Number of duplicates:", duplicates.sum())

# loại bỏ
final_df = final_df.drop_duplicates(subset=["Date", "ticker"])
```

```python
price_df = final_df.pivot(index="Date", columns="ticker", values="Close")
```

```python
returns = price_df.pct_change().dropna()
```

```python
mean_returns = returns.mean()
volatility = returns.std()

print("\n=== MEAN RETURNS ===")
print(mean_returns)

print("\n=== VOLATILITY ===")
print(volatility)
```

```python
corr_matrix = returns.corr()
print("\n=== CORRELATION MATRIX ===")
print(corr_matrix)
```

```python
cov_matrix = returns.cov()
print("\n=== COVARIANCE MATRIX ===")
print(cov_matrix)
```

```python
import numpy as np

num_assets = len(returns.columns)
weights = np.ones(num_assets) / num_assets  # chia đều

portfolio_return = np.dot(weights, mean_returns)
portfolio_vol = np.sqrt(np.dot(weights.T, np.dot(cov_matrix, weights)))

print("\n=== PORTFOLIO ===")
print("Weights:", weights)
print("Return:", portfolio_return)
print("Risk:", portfolio_vol)
```

```python
portfolio_returns = returns.dot(weights)

VaR_95 = np.percentile(portfolio_returns, 5)

print("\n=== VaR 95% ===")
print(VaR_95)
```

```python
risk_free_rate = 0.03 / 252

sharpe = (portfolio_return - risk_free_rate) / portfolio_vol

print("\n=== SHARPE RATIO ===")
print(sharpe)
```
